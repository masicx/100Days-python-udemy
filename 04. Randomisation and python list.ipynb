## Random module

```python
import random

random_integer = random.randint(1, 100)
print(random_integer)
```

```python
random_number_0_to_1 = random.random() * 10
print(random_number_0_to_1)
```

```python
random_float = random.uniform(1, 10)
print(random_float)
```

```python
random_heads_or_tails = random.randint(0,1)
if random_heads_or_tails == 1:
    print("Heads")
else:
    print("Tails")
```

## Who will pay the bill

```python
import random

friends = ["Alice", "Bob", "Charlie", "David", "Emanuel"]
random_choice = random.choice(friends)
print(random_choice)
```

## Nested Lists

```python
fruits = ["Strawberry", "Nectarine", "Apple", "Pear", "Grape"]
vegetables = ["Spinach", "Kale", "Tomato", "Celery", "Potato"]

dirty_dozen = [fruits, vegetables]
print(dirty_dozen)
```

## Final project

```python
choices = ["Rock", "Paper", "Scissors"]
user_choice = input("What do you choose? Type 0 for Rock, 1 for Paper, 2 for Scissors.\n")
user_choice = int(user_choice)
computer_choice = random.randint(0, 2)
print(f"You chose {choices[user_choice]}")
print(f"Computer chose {choices[computer_choice]}")
if user_choice == computer_choice:
    print("It's a draw")
elif (user_choice == 0 and computer_choice == 2) or (user_choice == 1 and computer_choice == 0) or (user_choice == 2 and computer_choice == 1):
    print("You win!")
else:
    print("You lose!")
```
