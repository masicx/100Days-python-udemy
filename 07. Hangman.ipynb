```mermaid
graph TD
    A[START] --> B(Generate a random word)
    B --> C(Generate as many blanks as letters in word)
    C --> D(Ask the user to guess a letter)
    D --> E{Is the guessed letter in the word?}
    E --Yes--> F(Replace the blank with the letter)
    E --No--> G(Lose a life)
    G --> J{Have they run out of lives?}
    F --> H{Are all the blanks filled?}
    J --Yes --> I
    J --No--> D
    H --No--> D
    H --Yes-->I(End)
    
```

````python
import random
from IPython.display import clear_output

word_list = ["aardvark", "baboon", "camel"]
chosen_word = random.choice(word_list)
guess_word = ["_"] * len(chosen_word)
lives = 6
game_over = False

while not game_over:
    clear_output(wait=True)
    was_correct = False
    print(f"Current guess word: {str.join(' ', guess_word)}")
    print(f"Lives remaining: {lives}")
    guess = input("Guess a letter: ").lower()
    for index in range(0, len(chosen_word)):
        if chosen_word[index] == guess:
            guess_word[index] = guess
            was_correct = True
    if not was_correct:
        lives -= 1
    
    game_over = "_" not in guess_word or lives == 0
    if game_over:
        print("Game over")
````
