```python
clean_data_frame = data_frame.dropna()
clean_data_frame.tail()
```

```python
max_id = clean_data_frame["Starting Median Salary"].idxmax()
print(max_id)
clean_data_frame["Undergraduate Major"].loc[max_id]
```

```python
mid_career_median = clean_data_frame["Mid-Career Median Salary"].max()
print(mid_career_median)
clean_data_frame["Undergraduate Major"].loc[clean_data_frame["Mid-Career Median Salary"] == mid_career_median]
```

```python
lowest_starting_median = clean_data_frame["Starting Median Salary"].min()
print(lowest_starting_median)
clean_data_frame["Undergraduate Major"].loc[clean_data_frame["Starting Median Salary"] == lowest_starting_median]
```

```python
# Sorting values and adding columns: Majors with the most potential vs lowest risk
spread_col =clean_data_frame["Mid-Career 90th Percentile Salary"] - clean_data_frame["Mid-Career 10th Percentile Salary"]
clean_data_frame.insert(1, "Spread", spread_col)
```

```python
clean_data_frame.head()
```

```python
low_risk = clean_data_frame.sort_values("Spread")
low_risk[["Undergraduate Major", "Spread"]].head()
```

```python
# Majors with the highest potential
highest_potential = clean_data_frame.sort_values("Mid-Career 90th Percentile Salary", ascending=False)
highest_potential[["Undergraduate Major", "Mid-Career 90th Percentile Salary"]].head()
```

```python
# Major with the greatest spread in salaries
highest_spread = clean_data_frame.sort_values("Spread", ascending=False)
highest_spread[["Undergraduate Major", "Spread"]].head()
```

```python
# Grouping and pivoting data with pandas
print(clean_data_frame.groupby("Group").count())
```

```python
clean_data_frame.groupby("Group").mean()
```
