# Nesting lists and dictionaries

```python
# Nested list in dictionary
travel_log = {
    "France": ["Paris", "Lille", "Dijon"],
    "Germany": ["Berlin", "Hamburg", "Stuttgart"]
}

print(travel_log["France"][1])
```

```python
travel_log = {
    "France": {
        "cities_visited": ["Paris", "Lille", "Dijon"],
        "total_visits": 12
    },
    "Germany": {
        "cities_visited": ["Berlin", "Hamburg", "Stuttgart"],
        "total_visits": 5
    }
}

print(travel_log["Germany"]["cities_visited"][2])
```

## Final project

```python
from IPython.display import clear_output

print("Welcome to the secret auction program.")

another_bidder = True
auction_dictionary = {}

while another_bidder:
    name = input("What is your name? ")
    bid = int(input("What is your bid? $"))
    auction_dictionary[name] = bid
    another_bidder = input("Are there any other bidders? Type 'yes' or 'no'. ").lower() == 'yes'
    clear_output()

highest_bid = 0
winner = ""

for bidder in auction_dictionary:
    if auction_dictionary[bidder] > highest_bid:
        highest_bid = auction_dictionary[bidder]
        winner = bidder

print(f"The winner is {winner} with a bid of ${highest_bid}.")
```
