## Functions with inputs

```python
def greet_with_name(name):
    print(f"Hello, {name}!")
    print(f"How do you do, {name}?")

greet_with_name("Alice")
```

## Functions with multiple parameters

```python
def greet_with(name, location):
    print(f"Hello, {name}!")
    print(f"What is it like in {location}?")

greet_with("Alice", "London")
greet_with(location="London", name="Alice")
```

## Caesar Cipher

```python
alphabet = "abcdefghijklmnopqrstuvwxyz"
def encode(plain_text, shift_amount):
    cipher_text = ""
    for letter in plain_text:
        
        if letter not in alphabet:
            cipher_text += letter
            continue

        position = alphabet.index(letter)
        new_position = (position + shift_amount) % 26
        cipher_text += alphabet[new_position]
    print(f"The encoded text is {cipher_text}")

def decode(cipher_text, shift_amount):
    decoded_text = ""
    for letter in cipher_text:

        if letter not in alphabet:
            decoded_text += letter
            continue

        position = alphabet.index(letter)
        new_position = (position - shift_amount) % 26
        decoded_text += alphabet[new_position]
    print(f"The decoded text is {decoded_text}")

end_program = False
while not end_program:
    operation = input("Type 'encode' to encrypt, type 'decode' to decrypt").lower()
    message = input("Type your message: ").lower()
    shift = int(input("Type the shift number: "))
    if operation == "encode":
        encode(plain_text=message, shift_amount=shift)
    elif operation == "decode":
        decode(cipher_text=message, shift_amount=shift)
    else:
        print("Invalid operation")
        end_program = True

    try_again = input("Type 'yes' if you want to go again. Otherwise type 'no'.").lower()
    if try_again == "no":
        end_program = True
```
