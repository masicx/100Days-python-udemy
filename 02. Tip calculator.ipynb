## Primitive data types

1. String

```python
# Subscripting
print("Hello"[0])
print("Hello"[-1]) # print the last letter of the string

# String
print("123" + "345") # concatenation of strings
```

2. Integer

```python
print(123 +345) # addition of numbers
print(123_456_789) # underscores can be used for readability in numbers
```

3. Float

```python
print(3.14159) # float number
```


4. Boolean

```python
print(True)
print(False)
```

## Type Error, Checking and Conversion

```python
# len(123) # TypeError: object of type 'int' has no len()
print(len("123")) # 3

# Type checking
print(type(123)) # <class 'int'>
print(type(3.14159)) # <class 'float'>
print(type("123")) # <class 'str'>
print(type(True)) # <class 'bool'>

# Type conversion or casting
print(int("123") + int("456"))
# float()
# str()
# bool()

print("Number of letters in your name: " + str(len(input("Enter your name: ")))) # TypeError: can only concatenate str (not "int") to str
```

## Mathematical operations

```python
print(123 + 456)  # 579
print(2 - 1)  # 1
print(3 * 3)  # 9
print(6 / 3)  # 2.0
print(6 // 3)  # 2 returns integer
print(2**3)  # 8
```

PEMDAS

## Number manipulation and F string in python

## Final project day 2

```python
print("Welcome to the tip calculator!.")
total_bill = float(input("What was the total bill? $"))
tip_percentage = int(input("What percentage tip would you like to give? 10, 12, or 15? ")) / 100
number_of_people = int(input("How many people to split the bill? "))
tip_amount = total_bill * tip_percentage
total_bill_with_tip = total_bill + tip_amount
bill_per_person = total_bill_with_tip / number_of_people
final_amount = round(bill_per_person, 2)
print(f"Each person should pay: ${final_amount}")
```
